```python
#Crear una función que determine si un usuario tiene la edad edad adecuada para jugar Roblox
def es_mayor(edad):
  if edad >= 18:
    return "Puedes jugar y acceder a las versiones de pago"
  elif edad >= 12:
    return "Puedes jugar"
  else:
    return "No puedes jugar"
print(es_mayor(edad))
```

```python
#Crear un bucle que imprima los numeros del 1 al 10
# En python existen 2 tipos de bucle 1 se llama FOR se usa cuando sabemos donde termina
# El segundo bucle se llama while se usa cuando no sabemos donde termina

for recorrer in range(1, 11):
  print(recorrer)
```
